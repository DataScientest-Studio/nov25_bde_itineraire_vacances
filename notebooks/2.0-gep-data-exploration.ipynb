```python
import plotly.express as px
import pandas as pd
import numpy as np
import plotly.graph_objects as go
```

```python
# Read the json file extracted by the EXTRACTION Notebook

json_file = "..\\data\\raw\\pois_Île-de-France_20251218_160712.json"


# Read json file
data = pd.read_json(json_file)

# Convert to DataFrame
df = pd.DataFrame(data)
df.shape
```

```python
# ===== 1. DISTRIBUTION BY CATEGORY =====

category_counts = df['category'].value_counts()

# Pie Chart
fig_pie = px.pie(
    values=category_counts.values,
    names=category_counts.index,
    color=category_counts.index,
    color_discrete_sequence=px.colors.qualitative.Set3,
    title="Distribution par Catégorie (%)"
)
fig_pie.show()

# Histogram
fig_bar = px.bar(
    x=category_counts.index,
    y=category_counts.values,
    text=category_counts.values,
    color=category_counts.index,
    color_discrete_sequence=px.colors.qualitative.Set3,
    labels={'x': 'Catégorie', 'y': 'Nombre de POIs'},
    title="Nombre de POIs par Catégorie"
)
fig_bar.update_traces(textposition="outside")
fig_bar.show()
```

```python
# ===== 2. MISSING VALUES =====

fields = ['name','lat','lon','address','city','phone','website','opening_hours','wikipedia']
missing_data = {}

for field in fields:
    if field in ['lat','lon']:
        missing_count = df[field].isna().sum()
    else:
        missing_count = (df[field] == '').sum() + df[field].isna().sum()
    missing_percentage = (missing_count / len(df)) * 100
    missing_data[field] = {'count': missing_count, 'percentage': missing_percentage}

fields_sorted = sorted(missing_data.keys(), key=lambda x: missing_data[x]['percentage'], reverse=True)
percentages = [missing_data[f]['percentage'] for f in fields_sorted]

fig_missing = px.bar(
    x=percentages,
    y=fields_sorted,
    orientation='h',
    text=[f"{pct:.1f}%" for pct in percentages],
    color=percentages,
    color_continuous_scale="Reds",
    labels={'x': 'Pourcentage (%)', 'y': 'Champ'},
    title="Valeurs Manquantes par Champ"
)
fig_missing.update_traces(textposition="outside")
fig_missing.show()
```

```python
# ===== 3. TEMPORAL ANALYSIS =====
from datetime import datetime

df['timestamp_dt'] = pd.to_datetime(df['timestamp']).dt.tz_localize(None)
now = datetime(2025, 12, 15)

# Regroupement par mois
df['year_month'] = df['timestamp_dt'].dt.to_period('M')
temporal_counts = df.group_by('year_month').size().sort_index()

# Histogramme temporel
fig_hist_time = px.bar(
    x=[str(period) for period in temporal_counts.index],
    y=temporal_counts.values,
    text=temporal_counts.values,
    labels={'x': 'Mois', 'y': 'Nombre de mises à jour'},
    title="Histogramme Temporel des Mises à Jour"
)
fig_hist_time.update_traces(textposition="outside")

# Complément calcul métrics pour le temps
timestamps = df['timestamp_dt'].sort_values()
oldest = timestamps.iloc[0]
newest = timestamps.iloc[-1]
median_date = timestamps.iloc[len(timestamps)//2]
avg_timestamp = pd.Timestamp(timestamps.astype(np.int64).mean())

days_diff = [(now - ts).days for ts in timestamps]
within_30 = sum(1 for d in days_diff if d <= 30)
within_90 = sum(1 for d in days_diff if 30 < d <= 90)
within_365 = sum(1 for d in days_diff if 90 < d <= 365)
over_365 = sum(1 for d in days_diff if d > 365)

# Texte des métriques avec <br> pour les sauts de ligne
metrics_text = (
    f"Métriques temporelles:<br>"
    f"Dernière MAJ: {newest.strftime('%d-%m-%Y')}<br>"
    f"Plus ancien: {oldest.strftime('%d-%m-%Y')}<br>"
    f"Médiane: {median_date.strftime('%d-%m-%Y')}<br>"
    f"Moyenne: {avg_timestamp.strftime('%d-%m-%Y')}<br><br>"
    f"< 30j: {within_30}<br>"
    f"30-90j: {within_90}<br>"
    f"90-365j: {within_365}<br>"
    f"> 365j: {over_365}"
)

# Annotation sous le titre
fig_hist_time.add_annotation(
    text=metrics_text,
    xref="paper", yref="paper",
    x=0.1, y=1.08,   # centré, juste sous le titre
    showarrow=False,
    font=dict(size=12),
    align="left",
    bgcolor="wheat",
    bordercolor="black"
)

fig_hist_time.show()
```

```python
# VErsion heatmap temporelle

df['year'] = df['timestamp_dt'].dt.year
df['month'] = df['timestamp_dt'].dt.month

heatmap_data = df.group_by(['year','month']).size().reset_index(name='count')

fig_heatmap = px.density_heatmap(
    heatmap_data,
    x="month", y="year", z="count",
    color_continuous_scale="Viridis",
    title="Heatmap des Mises à Jour (Année vs Mois)"
)
fig_heatmap.show()
```

```python
# ===== 4. GEOGRAPHIC DENSITY=====

# complément densité
# Area zone calculation
lats = df['lat'].values
lons = df['lon'].values

min_lat, max_lat = lats.min(), lats.max()
min_lon, max_lon = lons.min(), lons.max()

# Approximation: 1 degree ≈ 111 km
lat_dist = (max_lat - min_lat) * 111
lon_dist = (max_lon - min_lon) * 111 * np.cos(np.radians((min_lat + max_lat) / 2))
area_km2 = lat_dist * lon_dist if lat_dist > 0 and lon_dist > 0 else 1

density = len(df) / area_km2

# Density metrics
density_text = f"""
DENSITÉ GÉOGRAPHIQUE

Densité: {density:.2f} POIs/km²
Zone couverte: {area_km2:.2f} km²
Total POIs: {len(df)}

Dimensions:
• Lat: {min_lat:.4f} à {max_lat:.4f}
• Lon: {min_lon:.4f} à {max_lon:.4f}
• Distance N-S: {lat_dist:.2f} km
• Distance E-O: {lon_dist:.2f} km
"""
```

```python
# Heatmap des POIs
fig_hex = px.density_heatmap(
    df,
    x="lon", y="lat",
    nbinsx=60, nbinsy=60,
    color_continuous_scale="Hot",
    title="Carte Hexbin de Densité des POIs"
)
fig_hex.show()
```

```python
# Scatter avec taille proportionnelle
poi_counts = df.group_by(['lat','lon']).size().reset_index(name='count')

fig_scatter = px.scatter(
    poi_counts,
    x="lon", y="lat",
    size="count",
    color="count",
    color_continuous_scale="Viridis",
    title="POIs en Île-de-France (taille = densité locale)"
)
fig_scatter.show()
```

```python
from sklearn.cluster import DBSCAN

# Préparer les coordonnées
coords = df[['lat','lon']].values

# Clustering DBSCAN
db = DBSCAN(eps=0.01, min_samples=5).fit(coords)  
# eps = rayon de voisinage (ajuste selon ton échelle), min_samples = taille min cluster

df['cluster'] = db.labels_

# Scatter interactif
fig_cluster = px.scatter(
    df,
    x="lon", y="lat",
    color="cluster",
    hover_name="name",
    title="Clustering Géographique des POIs (DBSCAN)"
)

# Mettre les outliers (-1) en rouge
fig_cluster.update_traces(marker=dict(size=10))
fig_cluster.show()
```

```python
# ===== STATISTIQUES DÉTAILLÉES =====
print("\n" + "="*60)
print("RAPPORT D'ANALYSE DES POIs")
print("="*60)

print("\n1. DISTRIBUTION PAR CATÉGORIE:")
for cat, count in category_counts.items():
    pct = (count / len(df)) * 100
    print(f"   • {cat}: {count} POIs ({pct:.1f}%)")

print("\n2. VALEURS MANQUANTES:")
for field in fields_sorted:
    data_field = missing_data[field]
    print(f"   • {field}: {data_field['count']} manquant(s) ({data_field['percentage']:.1f}%)")

print("\n3. ANALYSE TEMPORELLE:")
print(f"   • Dernière mise à jour: {newest.strftime('%d/%m/%Y')}")
print(f"   • Plus ancienne: {oldest.strftime('%d/%m/%Y')}")
print(f"   • Médiane: {median_date.strftime('%d/%m/%Y')}")
print(f"   • Moyenne: {avg_timestamp.strftime('%d/%m/%Y')}")
print(f"   • < 30 jours: {within_30} POIs")
print(f"   • 30-90 jours: {within_90} POIs")
print(f"   • 90-365 jours: {within_365} POIs")
print(f"   • > 365 jours: {over_365} POIs")

print("\n4. DENSITÉ GÉOGRAPHIQUE:")
print(f"   • Densité: {density:.2f} POIs/km²")
print(f"   • Zone couverte: {area_km2:.2f} km²")
print(f"   • Total POIs: {len(df)}")

print("\n" + "="*60)
```
