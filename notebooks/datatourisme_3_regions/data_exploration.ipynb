```python
import pandas as pd
```

```python
poi_df= pd.read_csv("data/data-poi.csv", usecols=['poi_id', 'main_category_id', 'sub_category_id', 'adresse_id'])
main_cat_df= pd.read_csv("data/data-main-cat.csv")
sub_cat_df= pd.read_csv("data/data-sub-cat.csv")
address_df= pd.read_csv("data/data-address.csv", 
                        usecols=['id', 'commune', 'code_postal'],
                        )
region_df=pd.read_csv("data/communes-france-2025.csv", 
                      usecols=['nom_standard', 'reg_nom', 'dep_nom', 'dep_code'], 
                      dtype={'nom_standard': str, 'reg_nom': str, 'dep_nom': str, 'dep_code': str})
```

```python
address_df['dep_code']=address_df['code_postal'].apply(lambda x: x[:2], axis= 0)
```

```python
df= poi_df.merge(main_cat_df, how='left', left_on='main_category_id', right_on='id')\
    .merge(sub_cat_df, how='left', left_on='sub_category_id', right_on='id')\
    .merge(address_df, how='left', left_on='adresse_id', right_on='id')\
    .merge(region_df, how='left', left_on='commune', right_on='nom_standard')
df= df[['poi_id', 'nom_cat', 'nom_sous_cat', 'commune', 'dep_code', 'dep_nom', 'reg_nom']]
df.info()
```

```python
poi_by_dept = df.groupby(['dep_code', 'dep_nom']).size().reset_index(name='nombre_poi')
```

```python
import plotly.express as px

poi_sorted = poi_by_dept.sort_values('nombre_poi', ascending=True)

fig = px.bar(poi_sorted, 
    x='nombre_poi', 
    y='dep_nom',
    orientation='h',
    title='Nombre de POI par département',
    labels={'nombre_poi': 'Nombre de POI', 'dep_nom': 'Département'},
    color='nombre_poi',
    color_continuous_scale='Viridis'
)

fig.show()
```
